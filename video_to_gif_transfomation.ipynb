```python
import moviepy
```

```python
file_name = '2.LiDAR_map'
input_path = f"./{file_name}.mov"
output_path = f"{file_name}.gif"

# Load the input video file
video = moviepy.VideoFileClip(input_path).resized(0.5)

# Convert the video to a GIF and save it to the output file path
video.write_gif(output_path,fps=5)

# Close the video file
video.close()
```
